## Dictionary with values 

### using curly braces

```python
dict3 = {'TS':'Hyderabad','TN':'Chennai','KA':'Bengaluru','KL':'Thiruvananthapuram'}
type(dict3)
```

### using assignment operator(identifier)

```python
dict4 = dict(TS = 'Hyderabad',TN = 'Chennai',KA = 'Bengaluru',KL = 'Thiruvananthapuram')
type(dict4)
```

### using list of lists

```python
dict5 = dict([['TS','Hyderabad'],['TN','Chennai'],['KA','Bengaluru'],['KL','Thiruvananthapuram']])
type(dict5)
```

### using tuple of tuples

```python
dict6 = dict((('TS','Hyderabad'),('TN','Chennai'),('KA','Bengaluru'),('KL','Thiruvananthapuram')))
type(dict6)
```

### using zip function

```python
states = ['TS','TN','KA','KL']
capitals = ['Hyderabad','Chennai','Bengaluru','Thiruvananthapuram']
dict(zip(states,capitals))
```
