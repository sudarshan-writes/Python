```python
import cv2
import array
```

```python
input_data = []
output = []
path1 = r'C:\Users\gsudh\Downloads\data\data\class_10_train'
os.chdir(path1)
for folder in os.listdir():
    path2 = path1 + '\\'+folder+'\\'+'images'
    os.chdir(path2)
    for image in os.listdir():
        input_data.append(cv2.imread(image))
        output.append(classes.loc['class'][folder])
```

```python
X = np.array(input_data)/255
```

```python
X.shape
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
y = le.fit_transform(output)
```

```python
from tensorflow.keras.layers import InputLayer,Conv2D,MaxPool2D,Dense,Flatten
from tensorflow.keras.models import Sequential
```

```python
from tensorflow.keras.layers import InputLayer, Conv2D,MaxPool2D,Dense, Flatten
from tensorflow.keras.models import Sequential

#CNN Architecture
model = Sequential()
model.add(InputLayer(shape=(64,64,3)))
model.add(Conv2D(filters = 10,kernel_size=(3,3), strides=(1, 1),padding='valid',activation='relu',kernel_initializer='glorot_uniform'))
model.add(Conv2D(filters = 10,kernel_size=(3,3), strides=(1, 1),padding='valid',activation='relu',kernel_initializer='glorot_uniform'))
model.add(MaxPool2D( pool_size=(2,2)))

model.add(Conv2D(filters = 10,kernel_size=(3,3), strides=(1, 1),padding='valid', activation='relu',kernel_initializer='glorot_uniform'))
model.add(Conv2D(filters = 10,kernel_size=(3,3), strides=(1, 1),padding='valid',activation='relu',kernel_initializer='glorot_uniform'))
model.add(MaxPool2D( pool_size=(2,2)))

model.add(Flatten())


#ANN Architecture
model.add(Dense(units=64,activation='relu'))
model.add(Dense(units=32, activation='relu'))
model.add(Dense(units=16,activation='relu'))

model.add(Dense(units=10, activation='softmax'))

model.summary()
```

```python
model.compile(optimizer = 'adam', loss= 'sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X,y,epochs=100,validation_split=0.2)
```
