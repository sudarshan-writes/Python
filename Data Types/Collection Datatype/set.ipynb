### 3. set operations

```python
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}
```

```python
set1.intersection(set2)  #common elements
```

```python
set1.union(set2) #unites all values without duplicates
```

```python
set1.difference(set2) # all elements that are in set1 but not not in set2
```

```python
set2.difference(set1)
```

```python
set1.symmetric_difference(set2)
```

```python
set1.intersection_update(set2)
print(set2)
print(set1)
```
