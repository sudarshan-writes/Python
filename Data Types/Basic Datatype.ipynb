**Three are Same then what is the difference**

```python
num = '345'
print(num.isdecimal())
print(num.isdigit())
print(num.isnumeric())
```

```python
num = '2³'
print(num.isdecimal())
print(num.isdigit())
print(num.isnumeric())
```

**one is false okay {*It doesnot accept Superscript format*} ,but remaining two are Same then what is the difference**

```python
num = 'Ⅹ'
print(num.isdecimal()) 
print(num.isdigit())
print(num.isnumeric())
```

***isdecimal // isdigit*** --- *Purely for english*
