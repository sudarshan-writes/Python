```python
x = int(input('enter a number::'))
y = int(input('enter a number::'))

print(x+y)
try:
    print(x/y)
    print(x//y)
except:
    print("You can't divide with zero")
    
print(x-y)
print(x*y)
print(x**y)
```

```python
try:
    x = int(input('enter a number::'))
    y = int(input('enter a number::'))

    print(x/y)
    print(x//y)
except(ZeroDivisionError,ValueError)as msg:
    print(msg)
    print(x+y)
    print(x-y)
    print(x*y)
    print(x**y)
```

```python
try:
    x = int(input('enter a number::'))
    y = int(input('enter a number::'))

    print(x+y)
    print(x-y)
    print(x*y)
    print(x**y)
    print(x/y)
    print(x//y)
    
except(ZeroDivisionError) as msg: #this gives the original error message
    print(msg)
    
except(ValueError): #this gives the customized error message
    print("input should be integer only")

else:
    print('no error occured')

finally:
    print('The code execution is completed')
```

## bank balance()

```python
class Insufficient_Fund_Error(Exception):
    pass
```

```python
balance = 15000
amount = int(input('enter the amount required'))
if amount <= balance:
    balance = balance - amount
    print('Take cash')
    print('your account balance is',balance)
else:
    #print("insufficient funds")
#instead of printing i want to raise a error so,
    raise Insufficient_Fund_Error('You dont have enough fund to withdraw')
```

Problem Statement: Write a Python program that asks the user to enter a number. If the input is not a valid integer, raise a ValueError and display a helpful error message instead of crashing.

Purpose: This exercise introduces the fundamental try/except pattern in Python. Handling invalid user input gracefully is one of the most common real-world uses of exception handling, and mastering it is a key step toward writing robust, user-friendly programs.

Given Input: User enters "hello" when prompted for a number.

```python
try:
    a = int(input('enter a number'))
    print(f"You gave number as {a}")
except(ValueError):
    print('Only Accepts numbers')
```
