```python
#create seperate lists for each data type

tuple1 = (1,2,4.5,'hello',True,3,"world")

Int = []
Float = []
Complex = []
String = []
Bool = []

for i in tuple1:
    if type(i) == int:
        Int.append(i)
    elif type(i) == float:
        Float.append(i)
    elif type(i) == str:
        String.append(i)
    elif type(i) == bool:
        Bool.append(i)
    elif type(i) == float:
        Int.append(i)
    else:
        print('data type is not specified')

print(Int)
print(Float)
print(String)
print(Bool)
```

## Combination of Ternary operator and Comprehension 

*T-Opr*: **(True val)if(condition)else(false val)**

*compre*: **expression for var in seq**

*combination*: **expression for var in seq if(condition)**

```python
#print list of all even numbers from 10 to 20

list2 = []
for i in range(10,21):
    if i%2 ==0:
        list2.append(i)
print(list2)
```

```python
[i for i in range(10,21) if (i%2 == 0)]
```

```python
#print list of all squares of odd numbers from 30 to 40

list2 = []
for i in range(30,41):
    if i%2 != 0:
        list2.append(i**2)
        print(i)
print(list2)
```
