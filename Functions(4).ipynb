## Recursive functions

## Factorial 

```python
# regular method

n = int(input('enter a number'))
fact = 1
for i in range(1,n+1):
    fact = fact * i
print(fact)
```

```python
# module method

import math
math.factorial(5)
```

```python
#recursive method

def fact(n):
    if (n == 0) or ( n == 1):
        return 1
    else:
        result = n * fact(n-1)
        return result

fact(5)
```

## Fibonacci series

```python
def fib(n):
    if (n==0) or (n==1):
        return n
    else:
        result = fib(n-1) + fib(n-2)
        return result
```

```python
x = int(input('enter a number'))
for i in range(x):
    print(fib(i))
```
