# Membership Operator

* This Operator checks whether the Value / Data is present or not in the sequence data. It Can be applied only on Sequence / Iterable Data.
* Have 2 keywords:
   * in
   * not in
* Return Data-type: Boolean

```python
a = 34
3 in a 
```

**Reason** : As 34 is combined single number and you are searching for single integer

```python
a = [3,4]
3 in a 
```

```python
name = 'hello world'
'w' in name
```

```python
name = 'hello world'
'ld' in name 
```

```python
name = 'hello world'
'wrl' in name 
```

**Reason**: Must be in the same sequence as given

```python
name = 'hello world'
'ow' in name
```

**Reason**:In the source 'space' is addressed and it's also a character

```python
Area = ['Kothapet','Dilsukhnagar','Chaitanyapuri','Nagole','Vanasthalipuram']
'Nagole' in Area
```

```python
name = 'hello world'
'z' not in name
```

```python
Area = ['Kothapet','Dilsukhnagar','Chaitanyapuri','Nagole','Vanasthalipuram']
'Hayathnagar' not in Area
```

# Identity Operator

* It compares the address of two objects.
  * Keywords: is, is not
    * To find address - id()

```python
a = 10
b = 10 
a is b #internally it's checking address of both and comparing not by the values
```

```python
id(a)
```

```python
id(b)
```

```python
x = 100
y = 100
x is y
```
