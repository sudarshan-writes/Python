## Valid website checker



```python
web = input('enter website name')

if web.startswith(('www.','http:','https')):
    
    if web.endswith(('.com','.in','.net','info')):
        print('Valid website')
    
    else:
        print('In_valid website')

else:
    print('Not a valid website')
```

```python
#optimised version
web = input('enter website name')

if web.startswith(('www.','http:','https')) and web.endswith(('.com','.in','.net','info')):
    print('Valid website')
else:
    print('In_valid website')
```

## Grade calculator

```python
marks = int(input('enter your marks::'))

if (90 < marks <= 100):
    print('A+ Grade')
    
elif(75 < marks <=90):
    print('A Grade')
    
elif(60 < marks <=75):
    print('B Grade')
    
elif(45 < marks <=60):
    print('C Grade')
    
elif(35 < marks <=45):
    print('D Grade')
    
elif(marks<0) or (marks>100):
    print('enter valid marks')

else:
    print("Fail")
```

# Ternary Operator

*Syntax*: **True_value if condition else False_value**

```python
#Traditional method
num = int(input('enter a number'))

if num%2 == 0:
    print('even')
else:
    print('odd')
```

```python
#ternary method
num = int(input('enter a number'))

'even' if num%2 == 0 else 'Odd'
```
